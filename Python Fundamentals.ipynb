8. for Loop
 
A for loop is a control statement used to repeat a block of code for each item in a sequence or for a specific number of times.

```python
for number in range(0, 6):
    print(number)
```

9. while Loop
A while loop is a control statement that repeats a block of code as long as a specified condition is True.

Real-Life Example Imagine filling a water tank.

While the tank is not full, keep filling it.

When the tank becomes full, stop.

The loop continues until the condition becomes False

```python
tank = 0

while tank < 5:
    tank += 1
    print("Filling tank")

print("Tank is full")
```

10. Break
    
The break statement is used to immediately terminate a loop, even if the loop has not completed all its iterations.

```python
for i in range(1, 11): 
    if i == 5: 
        break 
    print(i)
```

11. Continue
The continue statement is used to skip the current iteration of a loop and continue with the next iteration

```python
for i in range(1, 11): 
    if i == 5: 
        continue  # Skip the rest of the loop when i is 5
    print(i)
```

12. Pass

The pass statement is a placeholder that does nothing. It is used when Python requires a statement, but no action needs to be performed

```python
for i in range(1,5): 
    if i == 3: 
        pass 
    print(i)
```

13. List
    
Definition A list is an ordered, mutable collection that can store multiple items. It allows duplicate values and items can be added, removed, or modified.

Ordered → Items keep their position. Mutable → Items can be changed after creation. Real-Life Example

List ([]) → Can add, remove, and change items.

```python
#Create a List
shopping = ["Milk", "Bread", "Eggs", "Rice"]
print(shopping)

#Add a New Item (append())
shopping = ["Milk", "Bread", "Eggs", "Rice"]
shopping.append("Butter")
print(shopping)


#Remove an Item (remove())
shopping = ["Milk", "Bread", "Eggs", "Rice"]
shopping.remove("Eggs")
print(shopping)

#Change an Item (using Index)
shopping = ["Milk", "Bread", "Eggs", "Rice"]
shopping[1] = "Brown Bread"
print(shopping)


#Access an Item
shopping = ["Milk", "Bread", "Eggs", "Rice"]
print(shopping[2])


#Find the Total Number of Items
shopping = ["Milk", "Bread", "Eggs", "Rice"]
print(len(shopping))


#Sort the List
shopping = ["Milk", "Bread", "Eggs", "Rice"]
shopping.sort()
print(shopping)
```

14. Tuple

A tuple is an ordered, immutable collection that stores multiple items. Once created, its values cannot be changed.

Ordered → Items keep their position. Immutable → Items cannot be modified. Real-Life Example

Think of your Date of Birth.

```python
months = (
    "January",
    "February",
    "March",
    "April",
    "May",
    "June"
)

# Print the first month
print(months[0])

# Print the last month
print(months[-1])

# Find the length of the tuple
print(len(months))

# Check whether "March" is in the tuple
print("March" in months)
```

17. List Comprehension

Definition List comprehension is a short and simple way to create a new list by writing the loop and optional condition in a single line.

25. Recursive Functions

Definition
A recursive function is a function that calls itself to solve a problem. It continues calling itself until a stopping condition (called the base case) is reached.

Real-Life Example
Imagine standing between two mirrors.

You see your reflection repeating again and again.

It continues until the reflections become too small to notice.

Similarly, a recursive function keeps calling itself until it reaches the stopping condition.

```python
def countdown(n):
    if n == 0:
        print("Done")
    else:
        print(n)
        countdown(n - 1)

countdown(5)
```

36. Generators

Definition
A generator is a special type of function that produces values one at a time instead of returning all values at once.

Real-Life Example
Imagine a ticket machine.

It gives one ticket each time you press the button instead of printing all tickets at once.

A generator works the same way.

```python
#Generator
def numbers():
    yield 1
    yield 2
    yield 3

for num in numbers():
    print(num)
```

37. yield


Definition
The yield keyword is used inside a generator function to return one value at a time while saving the function's state. The function resumes from where it stopped when called again.

Real-Life Example
Think of a TV series.

You watch one episode today and continue from the next episode tomorrow.

It doesn't start from Episode 1 every time.

Similarly, yield pauses and resumes the function.

```python
## yield
def count():
    yield 1
    yield 2

g = count()

print(next(g))
print(next(g))
```

 38. Decorators
Definition
A decorator is a function that adds extra functionality to another function without changing its original code.

Real-Life Example
Think of gift wrapping.

The gift remains the same, but wrapping adds decoration.

Similarly, a decorator adds extra behavior to a function

```python
def decorator(func):
    def wrapper():
        print("Before")
        func()
        print("After")
    return wrapper

@decorator
def greet():
    print("Hello")

greet()
```

39. Closures

Definition
A closure is a function that remembers the variables from its outer function even after the outer function has finished executing.

Real-Life Example
Imagine your mother gives you a lunch box before you leave for school.

Even after you leave home, you still have the lunch box.

Similarly, the inner function remembers data from the outer function.

```python
def outer(name):
    def inner():
        print(name)
    return inner

message = outer("Nanda")
message()
```

40. global and nonlocal

Definition
global allows a function to modify a variable defined outside all functions (global scope). nonlocal allows an inner function to modify a variable in its enclosing (outer) function.

Real-Life Example
Imagine a company.

global → Company-wide announcement. nonlocal → Department-level announcement.

A company announcement affects everyone.

A department announcement affects only that department.

```python
count = 0

def update():
    global count
    count += 1

update()

print(count)
```
