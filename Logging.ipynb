**Module 4 - LOGGING**

**What is Logging?**

Logging is the process of recording events, messages, warnings, and errors that occur while a program runs.

**Why is Logging used?**

Logging helps us:

debug programs
track program flow
record errors
monitor behavior
save information for later analysis
Why use logging instead of print()?
print() is simple output, but logging is better because it supports:

1. log levels

2. formatting

3. file storage

4. console output

5. rotating logs

6. professional debugging

```python
#Example
import logging

logging.warning("Program Started")

import logging

logging.info("User Logged In")
```

**print() vs Logging**

**print()**

Used for simple output and quick checks.

**logging**

Used for structured message tracking in real applications.

**Main difference**

print() shows output directly. Logging gives control over:

importance of messages

where messages go

how messages look

**Log Levels**

**DEBUG**

Detailed information for debugging.

**INFO**

Normal program events.

**WARNING**

Something unexpected happened, but the program can continue.

**ERROR**

A serious problem happened in one operation.

**CRITICAL**

A very serious problem that may stop the program.

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("This is a DEBUG message")
logging.info("This is an INFO message")
logging.warning("This is a WARNING message")
logging.error("This is an ERROR message")
logging.critical("This is a CRITICAL message")
```

**Formatter**

A formatter decides how the log message should look.

Example: 2026-07-10 15:10:00 | INFO | Application started

A formatter makes logs:

1. readable
2. consistent
3. useful for debugging

```python
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s | %(levelname)s | %(message)s",
    datefmt="%Y-%m-%d %H:%M:%S",
    force=True
)

logging.info("Application started")
logging.warning("This is a warning")
logging.error("This is an error")
```

**Handlers**

A handler decides where the log message goes.

**Console Handler**
    
Sends logs to the screen or notebook output.

**File Handler**
    
Sends logs to a file so they can be saved and checked later.

```python
import logging

logging.basicConfig(
    filename="app.log",
    level=logging.INFO
)

logging.info("Application Started")
```

```python
#console hander
import logging

console = logging.StreamHandler()

logger = logging.getLogger()

logger.addHandler(console)

logger.warning("Console Message")
```

**Rotating Logs**
    
Rotating logs automatically create backup files when a log file becomes too large.

**Why is this useful?**
    
It prevents:

1. huge log files
2. difficult file management
3. storage problems

```python
import logging
from logging.handlers import RotatingFileHandler

handler = RotatingFileHandler(
    "app.log",
    maxBytes=100,
    backupCount=2
)

logger = logging.getLogger()

logger.addHandler(handler)

logger.warning("Log Message")
```

**Best Practices**

1. Use logging instead of too many print statements
2. Use correct log levels
3. Use formatters
4. Store important logs in files
5. Use rotating logs for long-running programs
6. Keep messages clear and meaningful
7. Avoid logging sensitive data
