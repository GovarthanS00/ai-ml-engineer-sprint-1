2. Data Types

Common Python data types:

int
float
str
bool
list
tuple
set
dict
None

```python
# Numerical Data Types
age = 25  # int
pi = 3.14159  # float

# Text & Boolean
name = "Alice"  # str
is_active = True  # bool

# Sequence Types
fruits = ["apple", "banana", "cherry"]  # list (mutable)
coordinates = (10.0, 20.0)  # tuple (immutable)

# Set Type
unique_numbers = {1, 2, 3, 3, 4}  # set (duplicates removed: {1, 2, 3, 4})

# Mapping Type
person = {"name": "Alice", "age": 25}  # dict

# None Type
empty_value = None  # NoneType

# Displaying values and their corresponding data types
data = [age, pi, name, is_active, fruits, coordinates, unique_numbers, person, empty_value]
```

int (Integer): Whole numbers without a decimal point (e.g., 10, -5).

float (Floating-Point): Real numbers with a decimal point or exponential notation (e.g., 3.14, -0.5).

str (String): Sequence of characters wrapped in single, double, or triple quotes (e.g., "Hello").

bool (Boolean): Truth value representing either True or False.

list: Ordered, mutable (changeable) collection that allows duplicate elements, written with square brackets [].

tuple: Ordered, immutable (unchangeable) collection that allows duplicate elements, written with parentheses ().

set: Unordered, mutable collection of unique elements (no duplicates), written with curly braces {}.

dict (Dictionary): Unordered collection of key-value pairs where keys must be unique, written with {key: value}.

NoneType (None): Represents the explicit absence of a value or a null value in Python.

4. Operators

Operators are symbols that perform actions on values.They are used for calculations, comparisons, or checking conditions.

Some examples are:

Arithematic operators : (+) Add (-) Subtract (*)Multiply (/) Divide

7. if / elif / else

if, elif, and else help Python make decisions.

if checks the first condition.
elif checks another condition if the first one is false.
else runs when none of the conditions are true

real word example

Imagine traffic lights.

If the light is Green → Go
Else if the light is Yellow → Slow down
Else → Stop

```python
light = input("Enter light color: ")

if light == "Green":
    print("Go")
elif light == "Yellow":
    print("Slow down")
else:
    print("Stop")
```

8. for Loop

A for loop is a control statement used to repeat a block of code for each item in a sequence or for a specific number of times.

```python
for number in range(0, 6):
    print(number)
```

9. while Loop

A while loop is a control statement that repeats a block of code as long as a specified condition is True.

Real-Life Example
Imagine filling a water tank.

While the tank is not full, keep filling it.

When the tank becomes full, stop.

The loop continues until the condition becomes False

```python
tank = 0

while tank < 5:
    tank += 1
    print("Filling tank")

print("Tank is full")
```

10. Break

The break statement is used to immediately terminate a loop, even if the loop has not completed all its iterations.

```python
for i in range(1, 11): 
    if i == 5: 
        break 
    print(i)
```

11. Continue

The continue statement is used to skip the current iteration of a loop and continue with the next iteration

```python
for i in range(1, 11): 
    if i == 5: 
        continue  # Skip the rest of the loop when i is 5
    print(i)
```

12. Pass

The pass statement is a placeholder that does nothing. It is used when Python requires a statement, but no action needs to be performed

```python
for i in range(1,5): 
    if i == 3: 
        pass 
    print(i)
```

Collections

13. List

Definition
   A list is an ordered, mutable collection that can store multiple items. It allows duplicate values and items can be added, removed, or modified.

Ordered → Items keep their position. Mutable → Items can be changed after creation. Real-Life Example

List ([]) → Can add, remove, and change items.


```python
#Create a List
shopping = ["Milk", "Bread", "Eggs", "Rice"]
print(shopping)

#Add a New Item (append())
shopping = ["Milk", "Bread", "Eggs", "Rice"]
shopping.append("Butter")
print(shopping)


#Remove an Item (remove())
shopping = ["Milk", "Bread", "Eggs", "Rice"]
shopping.remove("Eggs")
print(shopping)

#Change an Item (using Index)
shopping = ["Milk", "Bread", "Eggs", "Rice"]
shopping[1] = "Brown Bread"
print(shopping)


#Access an Item
shopping = ["Milk", "Bread", "Eggs", "Rice"]
print(shopping[2])


#Find the Total Number of Items
shopping = ["Milk", "Bread", "Eggs", "Rice"]
print(len(shopping))


#Sort the List
shopping = ["Milk", "Bread", "Eggs", "Rice"]
shopping.sort()
print(shopping)
```

14. Tuple
 
A tuple is an ordered, immutable collection that stores multiple items. Once created, its values cannot be changed.

Ordered → Items keep their position. Immutable → Items cannot be modified. Real-Life Example

Think of your Date of Birth.



Tuple (()) → Can only store, access, count, and search items. It cannot be modified after creation.

```python
months = (
    "January",
    "February",
    "March",
    "April",
    "May",
    "June"
)

# Print the first month
print(months[0])

# Print the last month
print(months[-1])

# Find the length of the tuple
print(len(months))

# Check whether "March" is in the tuple
print("March" in months)
```

17. List Comprehension

Definition
List comprehension is a short and simple way to create 
a new list by writing the loop and optional condition in a single line.

Real-Life Example
Imagine a teacher writing only the names of students who passed an exam.

Instead of writing a long process, the teacher directly creates a new list containing only the passed students.

Similarly, list comprehension creates a new list in one line.

```python
numbers = [1, 2, 3, 4, 5]

squares = [num ** 2 for num in numbers]

print(squares)
```

18. Dictionary Comprehension

Definition

Dictionary comprehension is a short and simple way to create a new dictionary by writing the loop and optional condition in a single line.

Real-Life Example
Imagine a teacher preparing a report card.

Instead of writing each student's name and mark one by one, the teacher quickly creates the entire report.

Similarly, dictionary comprehension creates key-value pairs efficiently.

```python
numbers = [1, 2, 3, 4]

square = {num: num ** 2 for num in numbers}

print(square)
```

19. Function

Definition

A function is a reusable block of code designed to perform a specific task. Instead of writing the same code multiple times, you can write it once inside a function and call it whenever needed.

20. Arguments

Definition

Arguments are the values passed to a function when it is called. They provide the information the function needs to perform its task.

Real-Life Example
Imagine ordering a pizza.

The pizza shop always asks:

What size? Which toppings?

The information you give is like arguments passed to the function.

```python
def greet(name):
    print("Hello", name)

greet("Arun")
greet("Bala")
```

21. Default Arguments

Default arguments allow a function to use a predefined value for a parameter if you don't provide one when calling it. If you pass a custom value, it overrides the default choice.

```python
def greet(name, city="Chennai"):
    print(name, city)

greet("Arun")
greet("Bala", "Coimbatore")
```

22. Keyword Arguments

Definition
Keyword arguments are arguments passed by parameter name, making it clear which value belongs to which parameter. The order of arguments doesn't matter.

Real-Life Example

Think of filling out an online application form.

Instead of writing values in order, you enter them under labeled fields like:

Name Age City

The labels identify the values.

```python
def student(name, age):
    print(name, age)

student(age=25, name="Nanda")
```
