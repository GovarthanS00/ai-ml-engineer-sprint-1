**5) PICKLE FILES**

A Pickle file is used to save Python objects directly into a file.

**Unlike JSON:**

JSON mainly stores common data types (dictionaries, lists, strings, numbers, etc.). Pickle can store almost any Python object.

**pickle.dump() → Python Object → Pickle File**

**pickle.load() → Pickle File → Python Object**

```python
#write 
import pickle

student = {
    "name": "Nanda",
    "age": 28
}

# Save
with open("student.pkl", "wb") as file:
    pickle.dump(student, file)

# Read
with open("student.pkl", "rb") as file:
    data = pickle.load(file)

print(data)
```

**7) PATHLIB**

**What is pathlib?**

pathlib is a modern Python module used to work with file and folder paths.

Instead of writing long file paths as strings, pathlib provides a cleaner, object-oriented way.
| Method      | Purpose                       |
| ----------- | ----------------------------- |
| `Path()`    | Create a path object          |
| `.exists()` | Check if a file/folder exists |
| `.name`     | Get the file name             |


```python
# Without pathlib
file = open("student.txt", "r")
print(file.read())
file.close()


# With pathlib
from pathlib import Path

file = Path("student.txt")

print(file.read_text())
```

```python
from pathlib import Path

file = Path("student.txt")

print(file.name)
```

**9) SHUTIL MODULE**
    
**What is shutil?**

shutil stands for:

Shell Utilities

It is used for high-level file and folder operations.

**With shutil, you can:**
    
Copy files Move files Copy folders Delete folders

Think of it as copy, paste, move, and delete in Windows Explorer.

                                           Function	                Purpose
                                        shutil.copy()	    Copy a file
                                        shutil.move()	    Move a file
                                        shutil.rmtree()	    Delete a folder and its contents


```python
import shutil

# Copy student.txt to backup.txt
shutil.copy("student.txt", "backup.txt")

print("File copied successfully!")

# Before:
# student.txt

# After:
# student.txt
# backup.txt
```

**10) FILE VALIDATION**
    
**What is File Validation?**

File Validation means:

Check whether a file or folder exists before using it.

```python
#with os 
import os

if os.path.exists("student.txt"):
    print("File exists")
else:
    print("File not found")
```

```python
##Using pathlib
from pathlib import Path

file = Path("student.txt")

if file.exists():
    print("File exists")
else:
    print("File not found")
```
