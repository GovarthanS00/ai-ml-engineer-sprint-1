**4.Foreign Key**
    
A foreign key is a column that links one table to another by referring to the primary key of another table.

**Example:**

```python
CREATE TABLE Department (
    DeptID INT PRIMARY KEY,
    DeptName VARCHAR(50)
);

INSERT INTO Department
VALUES
(1, 'HR'),
(2, 'IT'),
(3, 'Finance');

CREATE TABLE Employee (
    EmpID INT PRIMARY KEY,
    Name VARCHAR(50),
    DeptID INT,
    FOREIGN KEY (DeptID) REFERENCES Department(DeptID)
);

INSERT INTO Employee
VALUES
(101, 'Ram', 1),
(102, 'Priya', 2),
(103, 'Ravi', 3);

| EmpID | Name  | DeptID |
| ----: | ----- | -----: |
|   101 | Ram   |      1 |
|   102 | Priya |      2 |
|   103 | Ravi  |      3 |

Invalid Query
INSERT INTO Employee
VALUES (104, 'Kiran', 5);
Output
ERROR: Cannot add or update a child row:
a foreign key constraint fails
Key Point
Department is the Parent Table.
Employee is the Child Table.
DeptID in Employee is the Foreign Key.
It must match a DeptID in the Department table.
```

**6. Constraints**

Constraints are rules applied to table columns to ensure the data is accurate and valid.Constraints are applied while creating the table and inserting the values.

**Common Constraints:**

NOT NULL – The column cannot have an empty value.
UNIQUE – All values in the column must be different.
PRIMARY KEY – Uniquely identifies each row.
FOREIGN KEY – Connects one table to another.
CHECK – Ensures values meet a condition.
DEFAULT – Assigns a default value if no value is provided.
    
**Example:**

CREATE TABLE Students ( Student_ID INT PRIMARY KEY, Name VARCHAR(50) NOT NULL, Age INT CHECK (Age >= 18), City VARCHAR(30) DEFAULT 'Chennai' );

In this example,

PRIMARY KEY makes Student_ID unique.
    
NOT NULL ensures Name cannot be empty.
    
CHECK ensures Age is at least 18.
    
DEFAULT sets City to "Chennai" if no value is given.
    

```python
CREATE TABLE Employee (
    EmpID INT PRIMARY KEY,
    Name VARCHAR(50) NOT NULL,
    Email VARCHAR(100) UNIQUE,
    Salary INT CHECK (Salary > 0),
    City VARCHAR(50) DEFAULT 'Chennai',
    DeptID INT,
    FOREIGN KEY (DeptID) REFERENCES Department(DeptID)
);

Insert Valid Data
INSERT INTO Employee
(EmpID, Name, Email, Salary, DeptID)
VALUES
(101, 'Ram', 'ram@gmail.com', 50000, 1);
Output
Query OK, 1 row affected
Employee Table
EmpID	Name	Email	Salary	City	DeptID
101	Ram	ram@gmail.com	50000	Chennai	1

City becomes Chennai automatically because of the DEFAULT constraint.
##Types: PRIMARY KEY, FOREIGN KEY, NOT NULL, UNIQUE, CHECK, DEFAULT.
```

**Queries**

**SELECT**

SELECT is used to retrieve data from a table.

**Example:**

SELECT Name, Age

FROM Students;

This displays the Name and Age columns from the Students table.

```python
SELECT * FROM Employee;
```

**WHERE**

WHERE is used to filter rows based on a condition.

**Example:**

SELECT *

FROM Students

WHERE Age > 20;

This displays only students whose age is greater than 20.

```python
SELECT * FROM Employee
WHERE Salary > 50000;
```

**ORDER BY**
    
ORDER BY is used to sort the result in ascending (ASC) or descending (DESC) order.

**Example:**

SELECT *

FROM Students

ORDER BY Age ASC;

This sorts students by age from smallest to largest.

```python
SELECT * FROM Employee
ORDER BY Salary DESC;
```

**GROUP BY**

GROUP BY is used to group rows that have the same values in a column.

**Example:**

SELECT Department, COUNT(*)

FROM Employees

GROUP BY Department;

This shows the number of employees in each department.

```python
SELECT DeptID, COUNT(*)
FROM Employee
GROUP BY DeptID;
```

**HAVING**

HAVING is used to filter grouped data after using GROUP BY.

**Example:**

SELECT Department, COUNT(*)

FROM Employees

GROUP BY Department

HAVING COUNT(*) > 5;

This displays only departments that have more than 5 employees.

```python
SELECT DeptID, COUNT(*)
FROM Employee
GROUP BY DeptID
HAVING COUNT(*) > 2;
```

DISTINCT is used to return only unique values by removing duplicates.

**Example:**

SELECT DISTINCT City

```python
SELECT DISTINCT DeptID
FROM Employee;
```

**Joins**

**1.INNER JOIN**
  
Returns only the rows that have matching values in both tables that are combined through a common column.

**Example:**

SELECT Students.Name, Marks.Marks

FROM Students INNER JOIN Marks

ON Students.Student_ID = Marks.Student_ID;

Only matching records from both the tables are retreived


```python
SELECT *
FROM Employee
INNER JOIN Department
ON Employee.DeptID=Department.DeptID;
```

**2.LEFT JOIN**

Returns all rows from the left table and the matching rows from the right table. If there is no match, NULL is returned.

**Example:**

SELECT Students.Name, Marks.Marks

FROM Students LEFT JOIN Marks

ON Students.Student_ID = Marks.Student_ID;

All the students are shown, even if they have no marks

```python
SELECT *
FROM Employee
LEFT JOIN Department
ON Employee.DeptID=Department.DeptID;
```

**3.RIGHT JOIN**

Returns all rows from the right table and the matching rows from the left table. If there is no match, NULL is returned.

**Example:**

SELECT Students.Name, Marks.Marks

FROM Students RIGHT JOIN Marks

ON Students.Student_ID = Marks.Student_ID;

All records from the Marks table are shown, including Student_ID 104, which has no matching student.

```python
SELECT *
FROM Employee
RIGHT JOIN Department
ON Employee.DeptID=Department.DeptID;
```

**4.FULL JOIN**
    
Returns all rows from both tables. If there is no match, NULL is returned for the missing values.

**Example:**

SELECT Students.Name, Marks.Marks

FROM Students FULL JOIN Marks

ON Students.Student_ID = Marks.Student_ID;

Shows all records from both tables.

```python
SELECT *
FROM Employee
FULL OUTER JOIN Department
ON Employee.DeptID=Department.DeptID;
```

**5.CROSS JOIN**
    
Also called cartesian join, returns every possible combination of rows from both tables.

**Example:**

SELECT Students.Name, Marks.Marks

FROM Students

CROSS JOIN Marks;

Every student is combined with every marks record. If there are 3 students and 3 marks records, the result contains 9 rows.

```python
SELECT *
FROM Employee
CROSS JOIN Department;
```

**Advanced SQL**
 
**UNION**
  
UNION combines the results of two or more SELECT queries and removes duplicate rows.

CASE is used to apply conditions and return different values based on those conditions.

**Example:**

**Subqueries**

A subquery is a query written inside another SQL query.

**Example :**

SELECT Name

FROM Students

WHERE Marks = (

SELECT MAX(Marks)

FROM Students);
This displays the student who has the highest marks.

**Common Table Expressions (CTEs)**

A CTE is a temporary result set created using the WITH statement. It makes complex queries easier to read.

**Example:**

WITH HighMarks AS (

SELECT Name, Marks

FROM Students

WHERE Marks > 80)
SELECT *

FROM HighMarks;

This creates a temporary table called HighMarks and displays students who scored above 80.

**Views**

A view is a virtual table created from the result of a SQL query.

**Example:**

CREATE VIEW TopStudents AS

SELECT Name, Marks

FROM Students

WHERE Marks > 80;

To view the data

SELECT * FROM TopStudents;

This displays only students who scored above 80.

**Indexes**

An index is used to speed up data retrieval from a table.

**Example:**

CREATE INDEX idx_name

ON Students(Name);

This creates an index on the Name column, making searches faster.
    
**Window Functions**
    
Window functions perform calculations across a set of rows without grouping them.

**Example:**


SELECT Name,Marks,RANK() OVER (ORDER BY Marks DESC) AS Rank

FROM Students;

This displays each student's rank while still showing every row.

**ROW_NUMBER()**

ROW_NUMBER() assigns a unique sequential number to each row.

**Example:**


SELECT Name,ROW_NUMBER() OVER (ORDER BY Marks DESC) AS Row_Num

FROM Students;

Each row gets a unique number.
**RANK()**

RANK() assigns a rank to each row. If there is a tie, the same rank is assigned, and the next rank is skipped.

**Example:**



SELECT Name,Marks,RANK() OVER (ORDER BY Marks DESC) AS Rank

FROM Students;

Rank 3 is skipped because two students share rank 2.

**DENSE_RANK()**

DENSE_RANK() assigns a rank to each row. If there is a tie, the same rank is assigned, but no ranks are skipped.

**Example:**


SELECT Name,Marks,DENSE_RANK() OVER (ORDER BY Marks DESC) AS Dense_Rank

FROM Students;

No rank is skipped.

**LAG()**

LAG() returns the value from the previous row.

**Example:**

SELECT Name,Marks,LAG(Marks) OVER (ORDER BY Marks) AS Previous_Marks

FROM Students;

This shows the marks of the previous student.

**LEAD()**

LEAD() returns the value from the next row.

**Example:**


SELECT Name,Marks,LEAD(Marks) OVER (ORDER BY Marks) AS Next_Marks

FROM Students;

This shows the marks of the next student.
