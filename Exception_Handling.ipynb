Common Built-in Exceptions Exception Description

ZeroDivisionError Raised when dividing by zero 10 / 0

ValueError Raised when a function gets a valid type but an invalid value int("abc")

TypeError Raised when an operation is performed on incompatible data types "10" + 5

NameError Raised when a variable is not defined print(x)

IndexError Raised when a list index is out of range lst[10]

KeyError Raised when a dictionary key is not found d["age"]

AttributeError Raised when an object has no requested attribute "hello".append(5)

FileNotFoundError Raised when a file does not exist open("abc.txt")

ImportError Raised when a module cannot be imported import xyzmodule

ModuleNotFoundError Raised when a module is not found import mymodul

```python
def divide(a, b):
    try:
        assert b != 0, "b must not be zero"
        return a / b
    except AssertionError as e:
        print(e)
    except ZeroDivisionError:
        print("not divisible by zero")

print(divide(16, 4))
print(divide(35, 0))
```

```python
#custom
class InsufficientBalanceError(Exception):
    pass

balance = 5000
withdraw = int(input("Enter withdrawal amount: "))

try:
    if withdraw > balance:
        raise InsufficientBalanceError("Insufficient Balance")

    print("Transaction Successful")

except InsufficientBalanceError as e:
    print(e)
```
